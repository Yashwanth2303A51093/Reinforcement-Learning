```python

import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines and number of plays
M = 3
N_play = 10
epsilon = 0.2

# Part B – Initialize knowledge tables
# N = number of times each machine is selected
# R = total reward obtained from each machine
# Q = estimated average reward of each machine

N = {machine: 0 for machine in machines}
R = {machine: 0.0 for machine in machines}
Q = {machine: 0.0 for machine in machines}

print("Initial Tables")
print("N =", N)
print("R =", R)
print("Q =", Q)

# Part C – Modified Epsilon-Greedy Decision
for play in range(1, N_play + 1):

    # Generate random number between 0 and 1
    r = random.random()

    # Find machine with highest Q-value
    max_q = max(Q.values())

    # Handle ties properly
    highest_machines = [
        machine for machine in machines
        if Q[machine] == max_q
    ]

    highest_machine = random.choice(highest_machines)

    if r < epsilon:
        # Exploration:
        # Select randomly from machines except highest Q machine

        exploration_machines = [
            machine for machine in machines
            if machine != highest_machine
        ]

        selected_machine = random.choice(exploration_machines)
        decision = "Exploration"

    else:
        # Exploitation:
        # Select machine with highest Q-value
        selected_machine = highest_machine
        decision = "Exploitation"

    # Part D – Reward Collection
    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected machine:", selected_machine)

    # Get reward from user
    try:
        reward = float(input("Enter reward: "))
    except ValueError:
        print("Invalid input! Please enter a number.")
        continue

    # Part E – Q-Value Update
    N[selected_machine] += 1
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", {m: round(Q[m], 2) for m in machines})


# Part F – Final Results
print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nMachine\tParticipation Count (N)\tTotal Reward (R)\tQ-Value")

for machine in machines:
    print(
        f"{machine}\t\t"
        f"{N[machine]}\t\t\t"
        f"{R[machine]:.2f}\t\t"
        f"{Q[machine]:.2f}"
    )

# Total reward received
total_reward = sum(R.values())

print("\nTotal Reward Received:", round(total_reward, 2))

# Best machine
best_machine = max(Q, key=Q.get)

print("Best Estimated Machine:", best_machine)
print("Best Q-Value:", round(Q[best_machine], 2))

```
