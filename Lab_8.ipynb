```python
GAMMA = 0.9

# Rewards
NORMAL_REWARD = -1
GOAL_REWARD = 10
OBSTACLE_REWARD = -5

# Policy
policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}

# Deterministic transitions
transitions = {
    ("S0", "right"): "S1",
    ("S0", "up"): "S0",       # boundary -> same state
    ("S0", "down"): "S0",     # obstacle -> same state
    ("S0", "left"): "S0",

    ("S1", "down"): "S3",
    ("S1", "up"): "S1",
    ("S1", "left"): "S0",
    ("S1", "right"): "G",

    ("S3", "right"): "S4",
    ("S3", "up"): "S1",
    ("S3", "down"): "S3",
    ("S3", "left"): "S3",

    ("S4", "up"): "G",
    ("S4", "down"): "S4",
    ("S4", "left"): "S3",
    ("S4", "right"): "S4"
}


def get_reward(next_state):
    if next_state == "G":
        return GOAL_REWARD
    elif next_state == "X":
        return OBSTACLE_REWARD
    else:
        return NORMAL_REWARD


def state_value_function(start_state):
    """
    Calculate V(s) using trajectory/path method.
    No Bellman equation is used.
    """

    state = start_state
    total_return = 0.0
    discount = 1.0

    while state != "G":

        action = policy[state]

        next_state = transitions[(state, action)]

        reward = get_reward(next_state)

        total_return += discount * reward

        discount *= GAMMA

        state = next_state

    return total_return


def action_value_function(state, action):
    """
    Calculate Q(s,a) using trajectory method.

    First take the given action.
    Then follow the policy until reaching the goal.
    """

    total_return = 0.0
    discount = 1.0

    # Take the specified action first
    next_state = transitions[(state, action)]
    reward = get_reward(next_state)

    total_return += discount * reward
    discount *= GAMMA

    state = next_state

    # Follow the policy afterwards
    while state != "G":

        action = policy[state]

        next_state = transitions[(state, action)]

        reward = get_reward(next_state)

        total_return += discount * reward

        discount *= GAMMA

        state = next_state

    return total_return


# -----------------------------------
# Display State Values
# -----------------------------------

states = ["S0", "S1", "S3", "S4"]

print("STATE VALUES")
print("-" * 30)

for state in states:
    value = state_value_function(state)
    print(f"V({state}) = {value:.3f}")


# -----------------------------------
# Display Action Value
# -----------------------------------

print("\nACTION VALUE")
print("-" * 30)

q_value = action_value_function("S0", "up")

print(f"Q(S0, Up) = {q_value:.3f}")
```
